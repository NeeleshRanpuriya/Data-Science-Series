```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
delivery=pd.read_csv('deliveries.csv')
captain=pd.read_csv('Player_Match.csv')
player=pd.read_csv('Player.csv')
```

```python
temp_df=player.merge(captain, on='Player_Id')[['Player_Name', 'Match_Id', 'Is_Captain']]
temp_df
```

```python
delivery=delivery.merge(temp_df, left_on=['match_id','batter'], right_on=['Match_Id','Player_Name'], how='left').fillna(0)
```

```python
delivery
```

```python
runs=delivery.groupby(['match_id','batter'])['batsman_runs'].sum().reset_index()
balls=delivery.groupby(['match_id','batter'])['batsman_runs'].count().reset_index()
sixes=delivery.query('batsman_runs == 6').groupby(['match_id','batter'])['batsman_runs'].count().reset_index()
fours=delivery[delivery['batsman_runs']==4].groupby(['match_id','batter'])['batsman_runs'].count().reset_index()
```

```python
final_df=runs.merge(balls, on=['match_id','batter'],suffixes=('_runs','_balls')).merge(fours, on=['match_id','batter'], how='left').merge(sixes, on=['match_id','batter'] ,how='left')
```

```python
final_df.fillna(0, inplace=True)
```

```python
final_df
```

```python
final_df.rename(columns={'batsman_runs_runs':'runs','batsman_runs_balls':'balls','batsman_runs_x':'fours','batsman_runs_y':'sixes'}, inplace=True)
```

```python
final_df['sixes']=final_df['sixes'].astype(np.int16)
final_df['fours']=final_df['fours'].astype(np.int16)
final_df['Is_Captain']=final_df['Is_Captain'].astype(np.int16)
```

```python
final_df['sr']=round(final_df['runs']/final_df['balls']*100, 2)
```

```python
# final_df.drop(columns=['balls'], inplace=True)
```

```python
final_df=final_df.merge(temp_df, left_on=['match_id','batter'], right_on=['Match_Id','Player_Name'], how='left').drop(columns=['Match_Id','Player_Name']).fillna(0)
```

```python
def Dream11_Points(row):
    points = 0
    points += row['runs'] + row['fours'] + 2 * row['sixes']

    if row['runs'] >= 100:
        points += 16
    elif row['runs'] >= 50 and row['runs'] < 100:
        points += 8
    elif row['runs'] >= 30 and row['runs'] < 50:
        points += 4
    elif row['runs'] == 0:
        points -= 2

    if row['balls'] >= 10:
        if row['sr'] >= 170:
            points += 6
        elif row['sr'] >= 150 and row['sr'] < 170:
            points += 4
        elif row['sr'] >= 130 and row['sr'] < 150:
            points += 2
        elif row['sr'] >= 60 and row['sr'] < 70:
            points -= 2
        elif row['sr'] >= 50 and row['sr'] < 60:
            points -= 4
        elif row['sr'] < 50:
            points -= 6

    return points
```

```python
final_df['points'] = final_df.apply(Dream11_Points, axis=1)
expo_final_df = final_df.sort_values(by='points', ascending=False)[['match_id','batter','points']]
```
